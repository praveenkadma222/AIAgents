```python
# You are building a smart kitchen appliance system. Every appliance in the kitchen 
# must follow a strict rule: It must have a way to start_cooking() and a way to stop_cooking()
# create an abstract base class kitchenAppliances with these two abstractmethod and 
# implement them in two child classes : microwave and airfryer

from abc import ABC, abstractmethod
class KitchenApp(ABC):
    @abstractmethod
    deb start_cooking(self) :
    """Start cooking """
    pass
    @abstractmethod
    def stop_cooking(self)
    """stop cooking"""
    pass

class microwave(KitchenApp):

    def start_cooking(self):
        print("microwave : cooking")
    def stop_cooking(self)
        print("Microwave: stopped")    
class Airfryer(KitchenApp):

    def start_cooking(self):
        print("Airfryer : cooking")
    def stop_cooking(self)
        print("Airfryer: stopped")   

my1=Microwave()
my1.start_cooking()
my1.stop_cooking()
```
